```python
import pandas as pd
import os 
from dotenv import load_dotenv
from pathlib import Path
import json
```

```python
load_dotenv()

markets_json_path = Path(os.getenv("DATASETS_DIR")) / "raw" / "polymarket" / "markets.json"
```

```python
with open(markets_json_path, "r", encoding="utf-8") as f:
    market_data = json.load(f)
```

```python
markets_df = pd.json_normalize(market_data)
```
